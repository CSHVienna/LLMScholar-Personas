# Inter-Annotator Agreement — Ethnicity

Computes raw agreement, Cohen's κ, and Krippendorff's α between annotators on the ethnicity inference labels.

```python
import pandas as pd
from sklearn.metrics import cohen_kappa_score, confusion_matrix, ConfusionMatrixDisplay
import krippendorff
import matplotlib.pyplot as plt
```

```python
ANNOTATOR1_PATH = "../../../data/ethnicity_inference/manual_labels_annotator1.csv"
ANNOTATOR2_PATH   = "../../../data/ethnicity_inference/manual_labels_annotator2.csv"

annotator1 = pd.read_csv(ANNOTATOR1_PATH)[["sample_index", "manual_label"]].rename(columns={"manual_label": "label_annotator1"})
annotator2   = pd.read_csv(ANNOTATOR2_PATH)[["sample_index", "manual_label"]].rename(columns={"manual_label": "label_annotator2"})

df = annotator1.merge(annotator2, on="sample_index")

# Exclude items where either annotator skipped
df = df[(df["label_annotator1"] != "skip") & (df["label_annotator2"] != "skip")].copy()

print(f"Items after merging (excl. skips): {len(df)}")
print()
print("Annotator 1 label distribution:")
print(df["label_annotator1"].value_counts())
print()
print("annotator2 label distribution:")
print(df["label_annotator2"].value_counts())
```

```python
import sys
sys.path.insert(0, '../..')
from libs.metrics.agreement import compute_agreement, print_agreement_summary

stats = compute_agreement(df, 'label_annotator1', 'label_annotator2')
print_agreement_summary(stats)
```

```python
from libs.metrics.agreement import plot_confusion
from libs.visuals.constants import ETHNICITY_PLOT_ORDER

categories = [c for c in ETHNICITY_PLOT_ORDER if c in {*df['label_annotator1'].dropna().unique(), *df['label_annotator2'].dropna().unique()}]
fig, ax = plot_confusion(
    df, 'label_annotator1', 'label_annotator2',
    labels=categories,
    title='Confusion Matrix — Ethnicity\n(rows = Annotator 1, cols = Annotator 2)',
)
ax.set_xlabel('Annotator 2')
ax.set_ylabel('Annotator 1')
fig.show()
```
