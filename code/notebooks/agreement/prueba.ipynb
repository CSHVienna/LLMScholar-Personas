# Diagnóstico de baja concordancia en `manual_validation_20`

Investigamos por qué algunos requests tienen 0% / 60% / 75% de acuerdo entre la decisión automática del pipeline y la anotación manual. Hipótesis del usuario: **Jaro-Winkler sobre-matchea nombres genéricos/hallucinados**.

Plan:

1. Cargar el CSV chequeado + mergear `oa_status` desde `factuality_full.csv`. Reproducir números actuales.
2. Auditar discordancias `auto=found / manual=no` → traer `matched_name`, `gt_field`, scores de JW.
3. Investigar las filas con `*_auto` vacío.
4. Comparar tres criterios: `SS OR OA` (actual) vs `SS AND OA` vs `OA solo`.
5. Resumen + recomendación.

```python
# ── 1. Cargar el CSV chequeado y mergear oa_status ─────────────────────────
CHECKED_CSV = '/home/asanchez/code/asanchez/LLMScholar-Personas/data/annotator_agreement/manual_validation_20_checked.csv'
JOIN_KEYS = ['model','language','role','task','location','k','target','field','subfield','run_id','name','lastname']

df_filled = pd.read_csv(CHECKED_CSV, sep=';', dtype=str).fillna('')
print(f'Manual CSV: {len(df_filled)} filas, {df_filled["request_id"].nunique()} requests')

# Trayendo oa_status_auto desde full_fact (si no estaba)
if 'oa_status_auto' not in df_filled.columns:
    df_pipe = full_fact[JOIN_KEYS + ['oa_status']].copy()
    for k in JOIN_KEYS:
        df_pipe[k] = df_pipe[k].astype(str)
    df_pipe = df_pipe.dropna(subset=JOIN_KEYS).drop_duplicates(subset=JOIN_KEYS)
    df_filled = df_filled.merge(
        df_pipe.rename(columns={'oa_status': 'oa_status_auto'}),
        on=JOIN_KEYS, how='left',
    )
    df_filled['oa_status_auto'] = df_filled['oa_status_auto'].fillna('')
    print(f'  oa_status_auto rellenado en {(df_filled["oa_status_auto"]!="").sum()}/{len(df_filled)} filas')

df_filled[['request_id','name','lastname','author_status_auto','oa_status_auto',
           'found_in_ss_manual','found_in_oa_manual']].head(10)
```

```python
# ── 1b. Reproducir tabla por request (criterio actual: SS OR OA) ───────────
def truthy(v):
    return str(v).strip().lower() in ('yes','y','true','1','t')

def manual_found_any(r):
    ss, oa = r['found_in_ss_manual'], r['found_in_oa_manual']
    if not ss and not oa:
        return None
    return truthy(ss) or truthy(oa)

def auto_found_or(r):
    ss, oa = r.get('author_status_auto',''), r.get('oa_status_auto','')
    if not ss and not oa:
        return None
    return (ss == 'found') or (oa == 'found')

df_filled['manual_found_any'] = df_filled.apply(manual_found_any, axis=1)
df_filled['auto_found']       = df_filled.apply(auto_found_or,   axis=1)

both = df_filled[df_filled['manual_found_any'].notna() & df_filled['auto_found'].notna()]
print(f'AUTHOR FOUND concordance (criterio actual): '
      f'{(both["manual_found_any"]==both["auto_found"]).sum()}/{len(both)} = '
      f'{100*(both["manual_found_any"]==both["auto_found"]).mean():.1f}%')

by_req = both.groupby('request_id').apply(
    lambda g: pd.Series({
        'n_personas': len(g),
        'agree':      (g['manual_found_any']==g['auto_found']).sum(),
    }),
    include_groups=False,
)
by_req['pct'] = (100*by_req['agree']/by_req['n_personas']).round(1)
# Ordenar como pidió el usuario
by_req.index = by_req.index.astype(int)
by_req = by_req.sort_index()
print('\nPor request:')
print(by_req)
```

## 2. Auditar discordancias `auto=found / manual=no`

Para cada fila donde el pipeline dice `author_status_auto='found'` pero el revisor manual dice `found_in_ss_manual='no'`, mergeamos contra `factuality_full.csv` para traer:

- `matched_name`, `researcher_id`, `gt_field`, `gt_citations` — quién es la persona real con la que JW matcheó
- `match_score`, `matched_fields` — score total y campos que sumaron
- `sim_display_name`, `sim_first_name`, `sim_last_name`, `sim_dn_vs_last`, `sim_dn_vs_first` — similitudes individuales

Luego clasificamos en tres categorías con reglas simples sobre los `sim_*`.

```python
# ── 2a. Mergear las filas del CSV con factuality_full para traer detalles JW ──
JW_COLS = ['matched_name','researcher_id','gt_field','gt_citations',
           'match_score','matched_fields',
           'sim_display_name','sim_first_name','sim_last_name','sim_second_name',
           'sim_dn_vs_last','sim_dn_vs_first',
           'oa_display_name','oa_works_count','oa_cited_by_count']

df_jw = full_fact[JOIN_KEYS + JW_COLS].copy()
for k in JOIN_KEYS:
    df_jw[k] = df_jw[k].astype(str)
df_jw = df_jw.dropna(subset=JOIN_KEYS).drop_duplicates(subset=JOIN_KEYS)

df_aud = df_filled.merge(df_jw, on=JOIN_KEYS, how='left')
print(f'Filas con detalle de JW traído: {df_aud["matched_name"].notna().sum()}/{len(df_aud)}')
```

```python
# ── 2b. Filtrar discordancias auto=found, manual_ss=no, mostrar a quién matcheó JW ──
disc = df_aud[
    (df_aud['author_status_auto'] == 'found') &
    (df_aud['found_in_ss_manual'].str.lower() == 'no')
].copy()

print(f'Discordancias auto=found / manual_ss=no: {len(disc)} filas\n')

cols_show = ['request_id','name','lastname','matched_name','gt_field','gt_citations',
             'match_score','matched_fields',
             'sim_display_name','sim_first_name','sim_last_name','sim_dn_vs_last',
             'notes_manual']
pd.set_option('display.max_colwidth', 50)
pd.set_option('display.width', 200)
print(disc[cols_show].to_string(index=False))
```

```python
# ── 2c. Clasificar cada discordancia ──────────────────────────────────────
# Reglas:
#  - "homonimia": JW encontró un nombre que coincide bien (first + last >= 0.85
#    cada uno, o display_name >= 0.85). Es otra persona con field/afiliación
#    incompatible. Aquí JW no puede resolver por sí solo — necesita field-check.
#  - "JW espurio": llegó a 5+ puntos pero ningún sim individual fue >=0.85; suma
#    de matches parciales (típico: dn_vs_last + dn_vs_first juntan puntos
#    cuando los nombres están desordenados o son cortos).
#  - "nombre genérico hallucinado": apellido en una lista corta de nombres
#    obviamente inventados ("Doe", "Smith", "Johnson", "Mustermann", "Brown",
#    "Green", "Gray", "Blue", "Yamada" con nombre triple, etc.).

GENERIC_LAST = {'smith','johnson','brown','green','gray','mustermann','doe'}
def is_generic(name: str, lastname: str) -> bool:
    nm = str(name).lower()
    ln = str(lastname).lower()
    # apellido común y nombre simple
    if ln in GENERIC_LAST:
        return True
    # combinación "John Doe Smith", "Alice Green Yamada" — nombre con 2+ tokens
    if len(nm.split()) >= 2 and any(t in nm for t in ('doe','john','alice','bob','charlie','diana','ethan')):
        return True
    if nm == 'max' and ln in ('mustermann','muster'):
        return True
    return False

def categorize(r):
    if is_generic(r['name'], r['lastname']):
        return 'nombre_generico'
    dn = float(r['sim_display_name'] or 0)
    fn = float(r['sim_first_name']   or 0)
    ln = float(r['sim_last_name']    or 0)
    if dn >= 0.85 or (fn >= 0.85 and ln >= 0.85):
        return 'homonimia'
    return 'jw_espurio'

disc['categoria'] = disc.apply(categorize, axis=1)
print('Conteo por categoría:')
print(disc['categoria'].value_counts())
print()
print('Tabla completa con categoría:')
print(disc[['request_id','name','lastname','matched_name','gt_field',
            'sim_display_name','sim_first_name','sim_last_name','sim_dn_vs_last',
            'match_score','categoria']].to_string(index=False))
```

```python
# ── 3a. Filas con author_status_auto vacío ─────────────────────────────────
empty_auto = df_filled[df_filled['author_status_auto'] == ''].copy()
print(f'Filas con author_status_auto vacío: {len(empty_auto)}\n')
print(empty_auto[['request_id','name','lastname','found_in_ss_manual',
                  'found_in_oa_manual','notes_manual']].to_string(index=False))
```

```python
# ── 3b. ¿Cuál es el author_status en factuality_full para esas mismas claves? ──
# Comparamos también el original_csv (no chequeado) vs factuality_full para distinguir
# entre "el dato nunca llegó al CSV" y "Excel lo borró".

ORIGINAL_CSV = '/home/asanchez/code/asanchez/LLMScholar-Personas/results/manual/manual_validation_20.csv'
df_orig = pd.read_csv(ORIGINAL_CSV, dtype=str).fillna('')

for _, r in empty_auto.iterrows():
    print(f'\n── request {r["request_id"]}: {r["name"]} {r["lastname"]} ──')

    # En el CSV original
    mask_orig = (df_orig['request_id'].astype(str) == str(r['request_id'])) & \
                (df_orig['name'] == r['name']) & (df_orig['lastname'] == r['lastname'])
    if mask_orig.any():
        orig_row = df_orig[mask_orig].iloc[0]
        print(f'  CSV original  : author_status_auto = {orig_row["author_status_auto"]!r}')
    else:
        print('  CSV original  : (no encontrada por nombre)')

    # En factuality_full (cast keys a str)
    mask_full = True
    for k in JOIN_KEYS:
        mask_full = mask_full & (full_fact[k].astype(str) == str(r[k]))
    sub = full_fact[mask_full]
    if len(sub):
        print(f'  factuality_full: author_status     = {sub["author_status"].iloc[0]!r}  '
              f'(matched_name={sub["matched_name"].iloc[0]!r})')
    else:
        print('  factuality_full: (no encontrada con esa combinación de keys)')

    # Encoding check del campo lastname
    enc_orig = r['lastname'].encode('utf-8')
    print(f'  bytes lastname (checked): {enc_orig!r}')
```

```python
# ── 4. Tres criterios de auto_found, concordancia global y por request ─────
def auto_found_and(r):
    ss, oa = r.get('author_status_auto',''), r.get('oa_status_auto','')
    if not ss or not oa:
        return None
    return (ss == 'found') and (oa == 'found')

def auto_found_oa(r):
    oa = r.get('oa_status_auto','')
    if not oa:
        return None
    return oa == 'found'

df_filled['auto_or']  = df_filled.apply(auto_found_or,  axis=1)
df_filled['auto_and'] = df_filled.apply(auto_found_and, axis=1)
df_filled['auto_oa']  = df_filled.apply(auto_found_oa,  axis=1)

summary = []
for label, col in [('A: SS OR OA (actual)', 'auto_or'),
                   ('B: SS AND OA',          'auto_and'),
                   ('C: OA solo',            'auto_oa')]:
    sub = df_filled[df_filled['manual_found_any'].notna() & df_filled[col].notna()]
    n = len(sub)
    ag = (sub['manual_found_any'] == sub[col]).sum()
    pct = 100*ag/n if n else float('nan')
    summary.append({'criterio': label, 'n_comparadas': n, 'agree': ag, 'pct': round(pct,1)})

print('Concordancia global por criterio:')
print(pd.DataFrame(summary).to_string(index=False))

# Por request, sólo para criterio C (OA solo) que es el más limpio
print('\nPor request, criterio C (OA solo):')
sub_c = df_filled[df_filled['manual_found_any'].notna() & df_filled['auto_oa'].notna()].copy()
sub_c['request_id'] = sub_c['request_id'].astype(int)
by_req_c = sub_c.groupby('request_id').apply(
    lambda g: pd.Series({
        'n_personas': len(g),
        'agree':      (g['manual_found_any']==g['auto_oa']).sum(),
    }),
    include_groups=False,
)
by_req_c['pct'] = (100*by_req_c['agree']/by_req_c['n_personas']).round(1)
print(by_req_c.sort_index())
```

```python
# ── 5. Resumen final ───────────────────────────────────────────────────────
print('═' * 70)
print('RESUMEN DE DIAGNÓSTICO')
print('═' * 70)

print(f'\nTotal filas anotadas: {len(df_filled)}')
print(f'Filas con auto vacío (excluidas del cálculo): {(df_filled["author_status_auto"]=="").sum()}')

print('\n── Discordancias auto=found / manual_ss=no por categoría ──')
print(disc['categoria'].value_counts().to_string())

print('\n── Concordancia bajo tres criterios ──')
print(pd.DataFrame(summary).to_string(index=False))

print('\n── Recomendaciones (para discutir antes de implementar) ──')
print("""
1. SS sobre-matchea: confirmar revisando la tabla de la sección 2c. Si la
   mayoría de discordancias caen en "nombre_generico" + "jw_espurio", subir
   `min_matches` de 5 a 6 o exigir `sim_display_name >= 0.85` como condición
   necesaria.

2. Homonimia: para esos casos, JW sí encontró a una persona real con ese
   nombre — el problema es field/afiliación. La solución es exigir overlap
   field_status='field_match' (o gt_field ≈ requested field) antes de marcar
   `author_status='found'`.

3. Filas con auto vacío: revisar sección 3 — si están en factuality_full,
   regenerar el CSV chequeado con encoding utf-8 explícito y separador `;`
   sin pasar por Excel; si NO están, hay un bug en la generación del CSV de
   validación (sección 3a/3b lo distingue).

4. Cambiar criterio OR → OA-solo o SS-AND-OA si la concordancia sube
   sustancialmente (sección 4); en ese caso, vale la pena re-pensar el rol
   del paso de JW como "candidate filter" más que como "ground truth".
""")
```

```python
# ── 6. ¿OA-JW fuzzy aporta o sólo agrega falsos positivos? ────────────────
# Estratificamos las filas OA-found por oa_match_score:
#   - exact = 1.0  → match contra display_name o alguno de los display_name_alternatives
#   - fuzzy < 1.0  → fallback JW (≥0.85) sobre los mismos candidatos
# Si la mayoría de las discordancias OA-found / manual_oa=no caen en el grupo fuzzy,
# JW dentro de OA está aportando ruido y se podría dejar sólo el match exacto.

df_oa = full_fact[JOIN_KEYS + ['oa_status','oa_match_score']].copy()
for k in JOIN_KEYS:
    df_oa[k] = df_oa[k].astype(str)
df_oa = df_oa.dropna(subset=JOIN_KEYS).drop_duplicates(subset=JOIN_KEYS)

df_aud2 = df_filled.merge(df_oa.rename(columns={'oa_match_score':'oa_match_score_full'}),
                          on=JOIN_KEYS, how='left', suffixes=('','_y'))

oa_found = df_aud2[df_aud2['oa_status_auto'] == 'found'].copy()
oa_found['oa_match_score_full'] = pd.to_numeric(oa_found['oa_match_score_full'], errors='coerce')
oa_found['oa_match_kind'] = oa_found['oa_match_score_full'].apply(
    lambda s: 'exact (1.0)' if s == 1.0 else ('fuzzy (JW <1.0)' if pd.notna(s) and s < 1.0 else 'unknown')
)
oa_found['manual_oa_yes'] = oa_found['found_in_oa_manual'].str.lower().eq('yes')

print('Tabla cruzada: tipo de match OA × decisión manual:')
ct = pd.crosstab(oa_found['oa_match_kind'], oa_found['manual_oa_yes'],
                 margins=True, margins_name='total')
ct.columns = [str(c) for c in ct.columns]
print(ct.to_string())

print('\nPrecision (manual_oa=yes | OA-found) por tipo de match:')
agg = oa_found.groupby('oa_match_kind').agg(
    n=('manual_oa_yes','size'),
    true_positives=('manual_oa_yes','sum'),
).assign(precision_pct=lambda d: (100*d['true_positives']/d['n']).round(1))
print(agg.to_string())

print('\nDetalle de OA-found / manual_oa=no:')
bad = oa_found[~oa_found['manual_oa_yes']][[
    'request_id','name','lastname','oa_match_score_full','oa_match_kind',
    'author_status_auto','notes_manual'
]]
print(bad.to_string(index=False))
```

```python
# ── 7. Simulación: ¿qué pasa si apagamos el JW fuzzy de OA? ───────────────
# Reglas: oa_status_auto = 'found' SOLO si oa_match_score == 1.0
# (es decir, match exacto contra display_name o display_name_alternatives).
# Cualquier match con score < 1.0 (que vino del fallback JW) lo bajamos a not_found.

# Hacemos un merge fresco para tener oa_match_score como número (no string)
df_sim = df_filled.merge(
    df_oa.rename(columns={'oa_match_score':'oa_score'}),
    on=JOIN_KEYS, how='left', suffixes=('','_y'),
)
df_sim['oa_score'] = pd.to_numeric(df_sim['oa_score'], errors='coerce')

# Nuevo oa_status simulado
def oa_status_exact_only(r):
    cur = r.get('oa_status_auto','')
    if cur == 'found' and (pd.isna(r['oa_score']) or r['oa_score'] < 1.0):
        return 'not_found'    # era found vía JW → degradamos
    return cur

df_sim['oa_status_exact'] = df_sim.apply(oa_status_exact_only, axis=1)

# Cuántas filas cambian
changed = (df_sim['oa_status_auto']=='found') & (df_sim['oa_status_exact']=='not_found')
print(f'Filas que cambian de found → not_found: {changed.sum()}')
print(df_sim[changed][['request_id','name','lastname','oa_score',
                       'found_in_oa_manual','found_in_ss_manual','notes_manual']]
      .to_string(index=False))
```

```python
# ── 7b. Concordancia bajo cada criterio incluyendo OA-exact-only ──────────
def auto_oa_exact(r):
    s = r.get('oa_status_exact','')
    if not s:
        return None
    return s == 'found'

def auto_or_exact(r):
    """SS-found OR OA-exact-found"""
    ss = r.get('author_status_auto','')
    oa = r.get('oa_status_exact','')
    if not ss and not oa:
        return None
    return (ss == 'found') or (oa == 'found')

def auto_and_exact(r):
    """SS-found AND OA-exact-found"""
    ss = r.get('author_status_auto','')
    oa = r.get('oa_status_exact','')
    if not ss or not oa:
        return None
    return (ss == 'found') and (oa == 'found')

df_sim['auto_oa_exact']  = df_sim.apply(auto_oa_exact,  axis=1)
df_sim['auto_or_exact']  = df_sim.apply(auto_or_exact,  axis=1)
df_sim['auto_and_exact'] = df_sim.apply(auto_and_exact, axis=1)
df_sim['manual_found_any'] = df_sim.apply(manual_found_any, axis=1)

summary2 = []
criteria = [
    ('A : SS OR OA (actual)',             'auto_or'),
    ('B : SS AND OA (actual)',            'auto_and'),
    ('C : OA solo (actual)',              'auto_oa'),
    ('D : OA exact-only',                 'auto_oa_exact'),
    ('E : SS OR OA-exact-only',           'auto_or_exact'),
    ('F : SS AND OA-exact-only',          'auto_and_exact'),
]
# Recalcular las columnas legacy en df_sim
df_sim['auto_or']  = df_sim.apply(auto_found_or,  axis=1)
df_sim['auto_and'] = df_sim.apply(auto_found_and, axis=1)
df_sim['auto_oa']  = df_sim.apply(auto_found_oa,  axis=1)

for label, col in criteria:
    sub = df_sim[df_sim['manual_found_any'].notna() & df_sim[col].notna()]
    n  = len(sub)
    ag = (sub['manual_found_any'] == sub[col]).sum()
    pct = 100*ag/n if n else float('nan')
    summary2.append({'criterio': label, 'n': n, 'agree': ag, 'pct': round(pct,1)})

print('Comparación de los 6 criterios:')
print(pd.DataFrame(summary2).to_string(index=False))

print('\nPor request bajo criterio D (OA exact-only):')
sub_d = df_sim[df_sim['manual_found_any'].notna() & df_sim['auto_oa_exact'].notna()].copy()
sub_d['request_id'] = sub_d['request_id'].astype(int)
by_d = sub_d.groupby('request_id').apply(
    lambda g: pd.Series({
        'n_personas': len(g),
        'agree':      (g['manual_found_any']==g['auto_oa_exact']).sum(),
    }),
    include_groups=False,
)
by_d['pct'] = (100*by_d['agree']/by_d['n_personas']).round(1)
print(by_d.sort_index())
```

## 5. Resumen y recomendación

Junta los tres hallazgos: categorías de discordancia, criterios de `auto_found`, y filas con auto vacío.

## 4. Comparar criterios `auto_found`

Si la hipótesis "JW sobre-matchea" es cierta, ignorar SS (o exigir SS AND OA) debería subir la concordancia.

- **A**: `SS_found OR OA_found` (actual)
- **B**: `SS_found AND OA_found`
- **C**: sólo `OA_found`

## 3. Filas con `*_auto` vacío

En el CSV chequeado, algunas filas tienen los `*_auto` vacíos (p. ej. Michael Mkhize, Jane Nkosi, Erika Muñoz). En el `manual_validation_20.csv` original esas mismas filas estaban etiquetadas como `hallucinated`. Causas posibles: encoding (carácter `ñ`), Excel/LibreOffice borrando campos al re-guardar con `;`, o falla de join contra `factuality_full`.
