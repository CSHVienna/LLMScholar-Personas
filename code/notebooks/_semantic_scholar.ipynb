```python
import pandas as pd
import glob
```

```python
FILES_PATTERN = '../../data/semantic_scholar_data/DataFrameRankings_Genderize_Namsor_*.csv'
files = glob.glob(FILES_PATTERN)
len(files)
```

```python
df = pd.DataFrame()
for fn in files:
    discipline = fn.split('_')[-1].replace('.csv','')
    _df = pd.read_csv(fn)
    df = pd.concat([df, _df], ignore_index=True)
    df.loc[:,'_discipline'] = discipline
```
